```python
import pandas as pd
import numpy as np
```

DATA INGESTION

```python
data = "UseCase - Airlines.xlsx"

flights = pd.read_excel(data, sheet_name="flights")
payment = pd.read_excel(data, sheet_name="payments")
bookings = pd.read_excel(data, sheet_name="bookings")
passengers = pd.read_excel(data, sheet_name="passengers")

print("Flights:", flights.shape)
print("Payment:", payment.shape)
print("Bookings:", bookings.shape)
print("Passengers:", passengers.shape)
```

DATA INSPECTION

```python
print("\nFlights:")
print(flights.head())

print("\nPayment:")
print(payment.head())

print("\nBookings:")
print(bookings.head())

print("\nPassengers:")
print(passengers.head())
```

```python
print("\nFlights Information:")
print(flights.info())

print("\nPayment Information:")
print(payment.info())

print("\nBookings Information:")
print(bookings.info())

print("\nPassenger Information:")
print(passengers.info())
```

```python
print("\nFlights:")
print(flights.dtypes)

print("\nPayment:")
print(payment.dtypes)

print("\nBookings:")
print(bookings.dtypes)

print("\nPassenger:")
print(passengers.dtypes)
```

VALIDATION

```python
flight_columns=flights.columns
booking_columns=bookings.columns
passenger_columns=passengers.columns
payment_columns=payment.columns

if (set(flight_columns) <= set(flights.columns) and 
    set(booking_columns) <= set(bookings.columns) and 
    set(passenger_columns) <= set(passengers.columns) and 
    set(payment_columns) <= set(payment.columns)): 
    print("\nSchemas verified successfully") 
else: print("\nSchema mismatch found")
```

DATA QUALITY 

```python
print("\nFlights")
print(flights.isnull().sum())

print("\nPayment")
print(payment.isnull().sum())

print("\nBookings")
print(bookings.isnull().sum())

print("\nPassengers")
print(passengers.isnull().sum())
```

```python
print("\nFlights")
print("Duplicate rows:", flights.duplicated().sum())

print("\nPayment")
print("Duplicate rows:", payment.duplicated().sum())

print("\nBookings")
print("Duplicate rows:", bookings.duplicated().sum())

print("\nPassengers")
print("Duplicate rows:", passengers.duplicated().sum())
```

```python
#Standardizing
passengers["first_name"] = (
    passengers["first_name"]
    .astype("string")
    .str.strip()
    .str.title()
)

passengers["last_name"] = (
    passengers["last_name"]
    .astype("string")
    .str.strip()
    .str.title()
)

flights["airline"] = (
    flights["airline"]
    .astype("string")
    .str.strip()
    .str.title()
)
```

CHECKING INVALID / UNKNOWN VALUES

```python
print("\nAirline Values") 
print(flights["airline"].value_counts(dropna=False)) 
print("\nBooking Status") 
print(bookings["status"].value_counts(dropna=False)) 
print("\nPayment Methods") 
print(payment["payment_method"].value_counts(dropna=False))
```

Removing duplicate flights

```python
flights = flights.drop_duplicates() 
print("\nFlights after removing duplicates:", flights.shape)
payment = payment.drop_duplicates()
print("Payment records after removing duplicates:", payment.shape)
bookings = bookings.drop_duplicates()
print("Booking records after removing duplicates:", bookings.shape)
passengers = passengers.drop_duplicates()
print("Passenger records after removing duplicates:", passengers.shape)
```

```python
#DUPLICATE FLIGHT IDs
print("\nDuplicate flight IDs before cleaning:")
print(flights["flight_id"].duplicated().sum())
flights = flights.drop_duplicates(
    subset=["flight_id"],
    keep="first")
print("Duplicate flight IDs after cleaning:",flights["flight_id"].duplicated().sum())

#DUPLICATE BOOKING IDs
print("\nDuplicate booking IDs before cleaning:")

print(bookings["booking_id"].duplicated().sum())
bookings = bookings.drop_duplicates(subset=["booking_id"],keep="first")
print("Duplicate booking IDs after cleaning:",bookings["booking_id"].duplicated().sum())

#DUPLICATE PASSENGER IDs
print("\nDuplicate passenger IDs before cleaning:")
print(passengers["passenger_id"].duplicated().sum())
passengers = passengers.drop_duplicates(subset=["passenger_id"],keep="first")
print("Duplicate passenger IDs after cleaning:",passengers["passenger_id"].duplicated().sum())

#DUPLICATE PAYMENT RECORDS
print("\nDuplicate payment rows:")
print(payment.duplicated().sum())
payment = payment.drop_duplicates()
```

HANDLING MISSING VALUES

```python
# Missing airline 
flights["airline"] = flights["airline"].fillna("UNKNOWN") 
# Missing booking status 
bookings["status"] = bookings["status"].fillna("UNKNOWN") 
# Missing passenger last name 
passengers["last_name"] = passengers["last_name"].fillna("UNKNOWN") 
```

Coverting Data types

```python
flights["departure_time"] = pd.to_datetime( flights["departure_time"], errors="coerce" ) 
flights["arrival_time"] = pd.to_datetime( flights["arrival_time"], errors="coerce" ) 
bookings["booking_date"] = pd.to_datetime( bookings["booking_date"], errors="coerce" ) 
passengers["date_of_birth"] = pd.to_datetime( passengers["date_of_birth"], errors="coerce" ) 
payment["amount"] = pd.to_numeric( payment["amount"], errors="coerce" )
```

Handling Overnight flights

```python
overnight = flights["arrival_time"] < flights["departure_time"] 
flights.loc[overnight, "arrival_time"] = ( flights.loc[overnight, "arrival_time"] + pd.Timedelta(days=1) )
```

Calculating flight duration

```python
flights["calculated_duration_hours"] = ( flights["arrival_time"] - flights["departure_time"] ).dt.total_seconds() / 3600 
flights["calculated_duration_hours"] = flights["calculated_duration_hours"].round(2)
print("\nCalculated flight duration:") 
print(flights[ ["flight_id", "departure_time", "arrival_time", "calculated_duration_hours"] ].head())
```

```python
flights = flights.drop(columns=["duration"],errors="ignore")
```

Checking the existing Duration

```python
if "duration" in flights.columns:
    flights["duration_timedelta"] = pd.to_timedelta(
        flights["duration"].astype(str),
        errors="coerce"
    )
else:
    flights["duration_timedelta"] = (
        flights["arrival_time"] - flights["departure_time"]
    )
flights["duration_hours"] = (flights["duration_timedelta"].dt.total_seconds() / 3600)
flights["duration_difference"] = (flights["duration_hours"] - flights["calculated_duration_hours"])
print("\nDuration comparison:")
print(flights[["flight_id", "duration_hours", "calculated_duration_hours", "duration_difference"]].head())
```

Validating Table Relationships

```python
invalid_booking_flights = bookings[ ~bookings["flight_id"].isin(flights["flight_id"]) ] 
print( "\nBookings with invalid flight IDs:", len(invalid_booking_flights) ) 
invalid_booking_passengers = bookings[ ~bookings["passenger_id"].isin( passengers["passenger_id"] ) ] 
print( "Bookings with invalid passenger IDs:", len(invalid_booking_passengers) )
invalid_payment_bookings = payment[ ~payment["booking_id"].isin( bookings["booking_id"] ) ] 
print( "Payments with invalid booking IDs:", len(invalid_payment_bookings) )
```

Handling Invalid Bookings

```python
invalid_bookings = bookings[ bookings["status"] == "INVALID" ].copy() 
valid_bookings = bookings[ bookings["status"] != "INVALID" ].copy() 
print("\nInvalid bookings:", len(invalid_bookings)) 
print("Valid bookings:", len(valid_bookings))
```

PII

```python
passengers_analysis = passengers.drop( columns=["first_name","last_name","aadhaar_id", "phone", "email","date_of_birth"], errors="ignore" ) 
bookings_analysis = valid_bookings.drop( columns=[ "passport_number", "emergency_contact_name", "emergency_contact_phone" ], errors="ignore" )

print("PII removed from analysis datasets.")

# print("Bookings columns:")
# print(bookings_analysis.columns)

# print("Passengers columns:")
# print(passengers_analysis.columns)
```

Flight KPIs

```python
average_duration = flights[ "calculated_duration_hours" ].mean() 
total_flights = flights["flight_id"].nunique() 
print("\nFlight KPIs") 
print("Total flights:", total_flights) 
print("Average flight duration:", round(average_duration, 2), "hours")
```

Traffic

```python
route_traffic = ( flights .groupby(["source", "destination"]) .size() .reset_index(name="flight_count") ) 
print("\nRoute-wise Traffic")
print(route_traffic)
```

```python
airline_distribution = ( flights .groupby("airline") .size() .reset_index(name="flight_count") ) 
print("\nFlights by Airline") 
print(airline_distribution)
```

Booking Status Analysis

```python
booking_status = ( valid_bookings .groupby("status") .size() .reset_index(name="booking_count") ) 
print("\nBooking Status") 
print(booking_status)
```

Payment Analysis

```python
payment_summary = ( payment .groupby("payment_method")["amount"] .agg(["count", "sum", "mean"]) .reset_index() ) 
print("\nPayment Summary") 
print(payment_summary)
```

Saving Cleaned Data

```python
flights.to_csv("cleaned_flights.csv", index=False) 
valid_bookings.to_csv( "cleaned_bookings.csv", index=False ) 
passengers_analysis.to_csv( "cleaned_passengers.csv", index=False ) 
payment.to_csv( "cleaned_payment.csv", index=False ) 
route_traffic.to_csv( "route_traffic.csv", index=False ) 
airline_distribution.to_csv( "airline_distribution.csv", index=False ) 
booking_status.to_csv( "booking_status.csv", index=False ) 
payment_summary.to_csv( "payment_summary.csv", index=False ) 
print("\nCleaned and aggregated files saved successfully.")
```
